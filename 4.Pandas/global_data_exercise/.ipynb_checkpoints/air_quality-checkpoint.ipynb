```python
import pandas as pd
```

```python
df = pd.read_csv("globalAirQuality.csv")
df.head(1)
```

```python
#Seleting data
df["country"].head()
df[["country", "aqi"]]

df.loc[0:2] 
df.iloc[0:3] # 3 -> Exclusive 

df.loc[0:1, ["city", "aqi"]]
df.iloc[0:2, [2, 11]] # int values only
df.iloc[0:2, 1:5 ]

# single scale value
# cell
df.at[0, "city"]
df.iat[0, 2]

# returns View -> be carefull while using pands 
# so will apply .copy()
```

```python
#Filtering
df[ (df["aqi"] > 100) & (df["temperature"] > 30) ]

df[ df["aqi"] > 100 ][["country", "city"]]
```

```python
# Query -- returns copy
# but for simplicity we use .copy()
aqi_val = 100
df.query("aqi > 100 & temperature > 30")[["city", "aqi"]]
df.query("aqi > @aqi_val & temperature > 30")[["city", "aqi"]]
```
