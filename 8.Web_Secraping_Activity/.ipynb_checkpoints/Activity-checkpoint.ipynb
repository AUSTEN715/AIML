```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
# Logic of downloading data 

# URL = "https://quotes.toscrape.com/page/1/"
page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("no valid pages anymore..")
        break
    with open(f"scrape_data/quotes{page_count}.html", "w") as f:
        f.write(res.text)
        print(f"downloaded from page {page_count}")

    page_count += 1
```

```python
# Extract Useful Info
info = []

for count in range(1, page_count):
        
    with open(f"scrape_data/quotes{count}.html", "r") as f:
        contents = f.read()
    
    soup = BeautifulSoup(contents, "lxml")

    # Important Logic
    all_quotes = soup.select("div.quote")
    for quote in all_quotes:
    
        tags = quote.select_one("div.tags")
        for tag in tags.select("a.tag"):
            if(tag.get_text(strip=True) == "life"):
                quote_by_author = quote.select("span.text")[0].get_text(strip=True)
                author = quote.select("small.author")[0].get_text(strip=True)
                info.append([author, quote_by_author])
```

```python
info_df = pd.DataFrame(info)
info_df
```
