```python
import requests
```

```python
URL = "https://www.scrapethissite.com/pages/simple"
res = requests.get(URL)
```

```python
print(res.status_code)
print(res.text) # HTML content
print(res.content) # Raw byte response
print(res.header) # Meta data
```

```python
with open("scraped_data/data1.html", "w") as f:
    f.write(res.text)
```

```python
# BeautifulSoup

from bs4 import BeautifulSoup

with open("scraped_data/data1.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml") # lxml --> makes parsing faster (html and xml)
```

```python
all_h3 = soup.find_all("h3")

all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)
    # print(h3.find_parent("div")["class"])

    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    all_countries.append([name, population])
```

```python
all_countries
```

```python
import pandas as pd

df = pd.DataFrame(all_countries, columns=["Name", "Population"])
df
```

```python
df.to_csv("cleaned_data/data.csv", index=False)
```
